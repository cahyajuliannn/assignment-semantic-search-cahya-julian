## 1. Import Library & Load Knowledge Base

```python
import pandas as pd
import numpy as np
import faiss
from google import genai  # sesuaikan dengan library Gemini API yang digunakan
from google.genai import types
import os

client = genai.Client(api_key=os.environ.get('GEMINI_API_KEY'))

df = pd.read_csv('../data/ecommerce_faq_knowledge_base.csv')
print('Total entri knowledge base:', len(df))
df.head()
```

## 2. Menyiapkan Unit Retrievable

Setiap entri FAQ (question + answer) diperlakukan sebagai satu unit yang akan di-embed dan disimpan ke vector database. Gabungkan question dan answer menjadi satu teks representatif per entri.

```python
df['content'] = 'Q: ' + df['question'] + ' A: ' + df['answer']
documents = df['content'].tolist()
print('Contoh unit retrievable:')
print(documents[0])
```

## 3. Membuat Embeddings

Ubah setiap unit retrievable menjadi vektor menggunakan Gemini Embeddings API.

```python
def get_embedding(text):
    result = client.models.embed_content(
        model='gemini-embedding-2',  # sesuaikan versi model embedding
        contents=text
    )
    return result.embeddings[0].values

embeddings = [get_embedding(doc) for doc in documents]
embeddings = np.array(embeddings).astype('float32')
print('Shape embeddings:', embeddings.shape)
```

## 4. Menyimpan Embeddings ke Vector Database (FAISS)

```python
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

print('Jumlah vektor tersimpan di index:', index.ntotal)
```

## 5. Mengimplementasikan Semantic Search

Fungsi untuk mengambil beberapa entri FAQ paling relevan berdasarkan query pengguna.

```python
def semantic_search(query, top_k=3):
    query_embedding = np.array([get_embedding(query)]).astype('float32')
    distances, indices = index.search(query_embedding, top_k)

    results = []
    for rank, idx in enumerate(indices[0]):
        results.append({
            'rank': rank + 1,
            'question': df.iloc[idx]['question'],
            'answer': df.iloc[idx]['answer'],
            'category': df.iloc[idx]['category'],
            'distance': distances[0][rank]
        })
    return results

# Contoh percobaan
test_results = semantic_search('kenapa paket saya belum sampai?')
for r in test_results:
    print(r)
```

## 6. Mendesain Prompt dan Context

Susun prompt terstruktur (Persona, Task, Context, Format) yang menyisipkan hasil semantic search sebagai bagian Context, sebelum dikirim ke LLM.

```python
def build_prompt(query, search_results):
    context = '\n\n'.join(
        f"FAQ: {r['question']}\nJawaban: {r['answer']}" for r in search_results
    )
    prompt = f'''Kamu adalah asisten customer service e-commerce yang ramah dan membantu.

Tugas: Jawab pertanyaan pengguna HANYA berdasarkan informasi FAQ berikut. Jika informasi tidak ditemukan dalam FAQ, katakan bahwa kamu belum memiliki informasi tersebut dan sarankan menghubungi customer service.

Konteks FAQ relevan:
{context}

Pertanyaan pengguna: {query}

Jawaban:'''
    return prompt
```

## 7. Mengintegrasikan LLM

```python
def generate_answer(query, top_k=3):
    search_results = semantic_search(query, top_k=top_k)
    prompt = build_prompt(query, search_results)

    response = client.models.generate_content(
        model='gemini-3.6-flash',
        contents=prompt,
        config=types.GenerateContentConfig(temperature=0.3)
    )
    return response.text, search_results

answer, sources = generate_answer('apakah bisa bayar pakai cicilan?')
print('Jawaban LLM:')
print(answer)
print()
print('Sumber FAQ yang digunakan:')
for s in sources:
    print('-', s['question'])
```

### Catatan Analisis Relevansi

Pengujian dilakukan dengan 5 query: 2 query yang jelas ada di FAQ, 2 query parafrase/sinonim, dan 1 query di luar knowledge base.

| No | Jenis Query | Query | Sumber Teratas | Retrieval | Respons LLM |
|----|-------------|-------|----------------|-----------|-------------|
| 1 | Ada di FAQ (kata berbeda sedikit) | Bagaimana cara melacak status pesanan saya? | Bagaimana cara mengetahui status pesanan saya saat ini? | Tepat | Sesuai FAQ |
| 2 | Ada di FAQ (identik) | Metode pembayaran apa saja yang tersedia? | Metode pembayaran apa saja yang tersedia? | Tepat | Sesuai FAQ |
| 3 | Parafrase | bisa bayar dicicil per bulan? | Apakah saya bisa membayar dengan cicilan? | Tepat | Sesuai FAQ |
| 4 | Parafrase | barang yang saya terima rusak, uangnya bisa dikembalikan nggak? | Apa yang harus dilakukan jika paket diterima dalam kondisi rusak? | Relevan sebagian | Menjawab bagian yang ada, menolak bagian refund |
| 5 | Tidak ada di FAQ | Siapa CEO perusahaan ini? | Bagaimana cara mengetahui status pesanan saya saat ini? | Tidak relevan (wajar) | Menolak menjawab |

**1. Query yang berhasil di-retrieve dengan baik**
- Query 1 dan 2 berhasil mengambil FAQ yang tepat di peringkat teratas. Query 2 identik dengan pertanyaan di FAQ, sedangkan query 1 hanya berbeda redaksi ("melacak" vs "mengetahui", "saat ini" tidak disebut).
- Query parafrase 3 ("bisa bayar dicicil per bulan?") tetap menemukan FAQ "Apakah saya bisa membayar dengan cicilan?" meskipun kalimatnya berbeda dan memakai bahasa yang lebih santai. Ini menunjukkan embedding menangkap makna ("dicicil" ≈ "cicilan"), bukan sekadar mencocokkan kata kunci.
- Pada query 1-3, LLM menjawab sesuai isi FAQ tanpa menambah informasi di luar konteks.

**2. Query dengan hasil kurang relevan / sebagian relevan**
- Query 4 menggabungkan dua maksud: kondisi barang rusak dan pengembalian dana (refund). Retrieval mengambil FAQ tentang paket rusak di peringkat 1, yang sesuai dengan maksud pertama, tetapi FAQ tersebut tidak menjelaskan refund. Jawabannya jadi hanya menjawab sebagian pertanyaan. Kemungkinan penyebabnya adalah pertanyaan pengguna mencakup lebih dari satu topik, atau knowledge base belum punya FAQ khusus refund/pengembalian dana. Hal ini perlu dicek dengan melihat apakah FAQ retur/refund muncul di peringkat 2-3.

**3. Respons LLM saat informasi tidak tersedia**
- Pada query 5 ("Siapa CEO perusahaan ini?"), retrieval tetap mengembalikan FAQ teratas yang tidak berhubungan ("status pesanan"). Hal ini wajar karena FAISS selalu mengembalikan top-k hasil terdekat, seberapa pun jauhnya.
- Meskipun konteks yang diberikan tidak relevan, LLM tidak mengarang jawaban. LLM menyatakan belum memiliki informasi tersebut dan menyarankan menghubungi Customer Service. Ini sesuai instruksi prompt ("HANYA berdasarkan FAQ").
- Perilaku serupa juga terlihat pada query 4: LLM jujur menyebut bagian refund tidak ada di FAQ dan tidak berspekulasi, sambil tetap membantu dengan informasi yang tersedia.
- Artinya desain prompt (persona, batasan "hanya berdasarkan FAQ", dan instruksi fallback) efektif mencegah halusinasi.

**4. Kesimpulan dan saran perbaikan**
- Semantic search bekerja baik untuk query langsung maupun parafrase, dan LLM patuh pada batasan konteks.
- Retrieval tidak punya mekanisme untuk mendeteksi query di luar knowledge base. Saat ini pencegahan halusinasi sepenuhnya bergantung pada prompt. Saran: tambahkan threshold jarak (distance) sehingga query yang terlalu jauh dari semua FAQ langsung dijawab dengan pesan fallback tanpa memanggil LLM, sekaligus menghemat biaya API.
- Untuk query multi-topik seperti query 4, pertimbangkan menaikkan `top_k` atau memecah query menjadi beberapa sub-pertanyaan.
- Pertimbangkan menambah FAQ tentang refund/pengembalian dana jika memang kebijakannya ada, serta mengganti `IndexFlatL2` dengan cosine similarity (normalisasi vektor + `IndexFlatIP`) agar skor lebih mudah ditafsirkan.
